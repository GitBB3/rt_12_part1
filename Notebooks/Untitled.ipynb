# Statistical analysis - Paired T-Test

## Imports

```python
import pandas as pd
import numpy as np
```

## Read the data

```python
left_file='duration_left.csv'
right_file='duration_right.csv'
def read_targets(read_file):
    df = pd.read_csv(read_file)
    
    random_targets = [
        df['x'].tolist(),
        df['y'].tolist(),
        df['duration_sec'].tolist()
    ]
    
    return random_targets

# Exemple d'utilisation
duration_left = np.array(read_targets(left_file))
duration_right = np.array(read_targets(right_file))
#print(duration_left)
#print(duration_right)
```

## Compute the difference between the observations

```python
nb_exp = len(duration_left[2])
print(f"Nombre d'expériences: nb_exp = {nb_exp}")
DoF = nb_exp - 1
print(f"Degree of freedom: DoF = {DoF}")
d = duration_left[2] - duration_right[2]
print(f"Difference: d = {d}")
d_mean = np.mean(d)
print(f"Mean difference: d_mean = {d_mean}")
sd = np.std(d)
print(f"Standard deviation: sd = {sd}")
SE_d = sd / np.sqrt(nb_exp)
print(f"Standard error: SE_d = {SE_d}")
T = d_mean / SE_d
print(f"t-statistic: T = {T}")
```
